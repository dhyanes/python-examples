```python
# Take two integer values
a = 20
b = 3

# Arithmetic operations
print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)
print("Division:", a / b)
print("Remainder:", a % b)
print("Quotient:", a // b)
print("Power:", a ** b)
```

```python
# Take two integer values
a = 10
b = 20

# Comparison operations
print("a > b:", a > b)
print("a < b:", a < b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)
```

```python
# Take two integer values
a = 10
b = 20

# Equality operations
print("a == b:", a == b)
print("a != b:", a != b)

# Check return type
print(type(a == b))
print(type(a != b))
```

```python
# Declaring binary, octal, and hexadecimal values
bin_val = 0b1010        # binary (10 in decimal)
oct_val = 0o12          # octal (10 in decimal)
hex_val = 0xA           # hexadecimal (10 in decimal)

print("Binary value:", bin_val)
print("Octal value:", oct_val)
print("Hexadecimal value:", hex_val)

# Convert 9876 into different number systems
num = 9876

print("Binary of 9876:", bin(num))
print("Octal of 9876:", oct(num))
print("Hexadecimal of 9876:", hex(num))
```
