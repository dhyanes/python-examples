```python
my_str = "Although that way may not be obvious at first unless you're Dutch."

# Without begin, end, step
print(f"You have sliced: {my_str[:]}")

# Begin = 0, end = len(), no step
print(f"You have sliced: {my_str[0:len(my_str)]}")

# Without begin and end, but with step
print(f"You have sliced: {my_str[::1]}")

# With begin, end, and step
print(f"You have sliced: {my_str[0:len(my_str):1]}")

# Begin & end positive, step negative → empty string
print(f"You have sliced: {my_str[10:20:-1]}")

# Custom slicing
print(f"You have sliced: {my_str[0::2]}")

print(f"You have sliced: {my_str[::3]}")

# Reverse using only step
print(f"You have sliced: {my_str[::-1]}")

# Reverse using begin, end, step
print(f"You have sliced: {my_str[len(my_str)-1::-1]}")

# Custom reverse step
print(f"You have sliced: {my_str[::-3]}")

print(f"You have sliced: {my_str[len(my_str)-1::-3]}")

# Given slicing
print(my_str[10:17:-1])
# Output: '' (empty string)

# "yaw ta"
print(f"You have sliced: {my_str[14:8:-1]}")

# "ess you"
print(f"You have sliced: {my_str[63:55:-1]}")
```

```python
my_str = "Although 8 that way may not be obvious at first unless you're Dutch"

# Count of 't' using find()
count = 0
i = my_str.find('t')
while i != -1:
    count += 1
    i = my_str.find('t', i + 1)
print(count)

# Count of t using index 
count = 0
start = 0

while 't' in my_str[start:]:
    i = my_str.index('t', start)
    count += 1
    start = i + 1

print(count)
```

```python
user_str = input("Enter a string: ")

# Check conditions
if user_str.startswith(' ') and user_str.endswith(' '):
    result = user_str.strip()
elif user_str.endswith(' '):
    result = user_str.rstrip()
elif user_str.startswith(' '):
    result = user_str.lstrip()
else:
    result = user_str  # no spaces at start or end

print("Result:", result)
```

```python
my_str = input("Enter the sentence: ")
sub_str = input("Enter the sub string to which the index need to be found: ")

found = False

for i in range(len(my_str) - len(sub_str) + 1):
    match = True
    
    for j in range(len(sub_str)):
        if my_str[i + j] != sub_str[j]:
            match = False
            break
    
    if match:
        print("Substring found at index:", i)
        found = True
        break

if not found:
    print("Substring not found")
```

```python
my_str = input("Enter the sentence:")
sub_str = input("Enter the substring to replace: ")
new_str = input("Enter the new substring: ")

found = False

for i in range(len(my_str) - len(sub_str) + 1):
    match = True
    
    for j in range(len(sub_str)):
        if my_str[i + j] != sub_str[j]:
            match = False
            break
    
    if match:
        # Build replaced string
        result = ""
        
        # before match
        for k in range(i):
            result += my_str[k]
        
        # new string
        result += new_str
        
        # after match
        for k in range(i + len(sub_str), len(my_str)):
            result += my_str[k]
        
        print(result)
        found = True
        break

if not found:
    print("sub_str not found")
```

```python
my_str = "Peter Piper Picked A Peck Of Pickled Peppers."
sub_str = "Peck"

# Find sub string in the given string
index = -1
for i in range(len(my_str) - len(sub_str) + 1):
    match = True
    for j in range(len(sub_str)):
        if my_str[i + j] != sub_str[j]:
            match = False
            break
    if match:
        index = i
        break

if index != -1:
    total_len = len(my_str)
    sub_len = len(sub_str)

    stars = total_len - sub_len
    left = stars // 2
    right = stars - left

# build string with spaces 
    temp = ""

    for _ in range(left):
        temp += " "

    temp += sub_str

    for _ in range(right):
        temp += " "

# Rstrip and Lstrip uses
    left_spaces = len(temp) - len(temp.lstrip())
    right_spaces = len(temp) - len(temp.rstrip())

    # Step 4: rebuild using '*'
    result = ""

    for _ in range(left_spaces):
        result += "*"

    result += sub_str

    for _ in range(right_spaces):
        result += "*"

    print(result)

else:
    print("sub_str not found")
```
