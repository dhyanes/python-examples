```python
while True:
    print("\nSelect Operation:")
    print("1. Addition Enter 1")
    print("2. Subtraction Enter 2")
    print("3. Multiplication Enter 3")
    print("4. Division Enter 4")
    print("5: To divide and find quotient enter 5")
    print("6: To divide and find remainder enter 6")
    print("7: To find num1 to the power of num2 enter 7")
    print("8: To exit enter 8")

    choice = int(input("Enter choice: "))

    # Exit condition (like do-while break)
    if choice == 8:
        print("Thank you")
        break

    if choice <= 7:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

        if choice == 1:
            print("Result:", num1 + num2)

        elif choice == 2:
            print("Result:", num1 - num2)

        elif choice == 3:
            print("Result:", num1 * num2)

        elif choice == 4:
            if num2 != 0:
                print("Result:", num1 / num2)
            else:
                print("Error: Division by zero")

        elif choice == 5:
            if num2 != 0:
                print("Result:", num1 // num2)
            else:
                print("Error: Division by zero")

        elif choice == 6:
            if num2 != 0:
                print("Result:", num1 % num2)
            else:
                print("Error: Division by zero")

        elif choice == 7:
            print("Result:", num1 ** num2)

    else:
        print("Invalid Choice")
```

```python
# Program to check if a year is a leap year using nested if
year = int(input("Enter a year: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```

```python

x = input("Enter a sentence: ")

# Remove the period and split into words
words = x.replace('.', '').split()
print(words)

result = ""

for i in range(len(words)):
    if i == 0:
        result += words[i] + " "   # keep first word as is
    else:
        result += words[i].lower() + " "  # convert others to lowercase

print(result.strip())
```

```python
Q. Write a python program to find below output (implements rjust and
ljust) using loop:-
Input:- 'Peter Piper Picked A Peck Of Pickled Peppers.', sub_str =
'Peck',
Output:- '*********************Peck********************'
```

```python
my_str = "Peter Piper Picked A Peck Of Pickled Peppers."
sub_str = "Peck"

# Find sub string in the given string
index = -1
for i in range(len(my_str) - len(sub_str) + 1):
    match = True
    for j in range(len(sub_str)):
        if my_str[i + j] != sub_str[j]:
            match = False
            break
    if match:
        index = i
        break

if index != -1:
    total_len = len(my_str)
    sub_len = len(sub_str)

    stars = total_len - sub_len
    left = stars // 2
    right = stars - left

# build string with spaces 
    temp = ""

    for _ in range(left):
        temp += " "

    temp += sub_str

    for _ in range(right):
        temp += " "

# Rstrip and Lstrip uses
    left_spaces = len(temp) - len(temp.lstrip())
    right_spaces = len(temp) - len(temp.rstrip())

    # Step 4: rebuild using '*'
    result = ""

    for _ in range(left_spaces):
        result += "*"

    result += sub_str

    for _ in range(right_spaces):
        result += "*"

    print(result)

else:
    print("sub_str not found")
```

```python
Q. Write a python program to find below output using loop:-
Input:- 'This is Python class', sep = ' is',
Output:- ['This', 'is', 'Python class']
```

```python
my_str = "This is Python class"
sep = " is"

parts = my_str.split(sep)

result = []

# first part
result.append(parts[0])

# add separator word ('is')
result.append(sep.strip())

# remaining part
result.append(parts[1])

print(result)
```
